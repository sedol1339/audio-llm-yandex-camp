```python
import re
from pathlib import Path
import json
```

```python
data = json.loads(Path('datasets/ml.lecture_transcription/ml.lecture_transcription.json').read_text())
```

```python
for row in data['result']:
    text = row['transcription']
    while (match := re.search(r'\*([^\*]*)\/([^\*]*)\*', text)) is not None:
        start, end = match.span()
        start2, end2 = match.span(2)
        text = text[:start] + text[start2:end2] + text[end:]
    row['transcription_clean'] = text
    # if '*' in text:
    #     print(row['transcription'])
```

```python
Path('datasets/ml.lecture_transcription/ml.lecture_transcription.json').write_text(json.dumps(data, ensure_ascii=False, indent=4))
```
