```python
import re
from pathlib import Path
import json
```

```python
text = Path('манга.txt').read_text()
```

```python
lines = text.splitlines()
```

```python
for i, line in enumerate(lines):
    line = line.strip()
    while (match := re.search(r'\*([^\*]+)\*', line)) is not None:
        block = match.groups(1)[0]
        line = line[:match.start()] + block.capitalize() + '.' + line[match.end():]
        line = line.replace('!!', '!').replace('!!', '!').replace('!!', '!')
    # print(line) 
    lines[i] = line

text = '\n'.join(lines)   
```

```python
groups = text.split('\n\n')
text = '\n'.join(g.replace('\n', ' ') for g in groups)
```
